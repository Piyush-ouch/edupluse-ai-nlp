# Student Feedback Emotion Detection & Topic Modeling (NLP Subject Project)

This notebook demonstrates the complete end-to-end Natural Language Processing (NLP) pipeline implemented in **EduPulse NLP Studio**:
1. **Step-by-Step Text Preprocessing**: Contraction expansion, noise removal, tokenization, negation-preserving stopword filtering, morphological lemmatization, and N-gram extraction.
2. **Multi-Paradigm Emotion Detection**:
   - *Paradigm 1*: Affective Lexicon with Context-Window Valence Shifters & Boosters
   - *Paradigm 2*: Classical Statistical ML (`TF-IDF + Calibrated LinearSVC`)
   - *Paradigm 3*: Deep Neural Architecture (`Subword Character N-Grams + TruncatedSVD Contextual Embeddings + Multi-Layer Perceptron`)
   - *Paradigm 4*: Hybrid Transformer-Style Ensemble
3. **Unsupervised Topic Modeling Comparison**:
   - **LDA** (*Latent Dirichlet Allocation*)
   - **NMF** (*Non-Negative Matrix Factorization*)
   - **BERTopic-Style Semantic Clustering** (*Dense Embeddings + K-Means + Class-Based TF-IDF `c-TF-IDF`*)

```python
import os
import sys
import pandas as pd
import numpy as np

# Add backend directory to path
sys.path.append(os.path.abspath('../backend'))
from nlp_engine import StudentFeedbackNLPEngine, preprocess_text_detailed

dataset_path = os.path.abspath('../data/student_feedback_dataset.csv')
engine = StudentFeedbackNLPEngine(dataset_path)
print(f'Loaded {len(engine.df)} student feedback records.')
engine.df.head()
```

## 1. Inspect Step-by-Step NLP Preprocessing Pipeline

```python
sample_text = "The instructor doesn't explain how mathematical proofs connect to code, leaving students confused and frustrated!"
prep = preprocess_text_detailed(sample_text)
for k, v in prep.items():
    print(f'{k:18s}: {v}')
```

## 2. Emotion Detection Model Benchmarks (Classical vs. Neural vs. Ensemble)

```python
rows = []
for model_name, metrics in engine.model_metrics['models'].items():
    rows.append({
        'Model Architecture': model_name,
        'Accuracy (%)': metrics['accuracy'],
        'Macro Precision (%)': metrics['macro_precision'],
        'Macro Recall (%)': metrics['macro_recall'],
        'Macro F1 (%)': metrics['macro_f1'],
        'Latency (ms/doc)': metrics['inference_latency_ms']
    })
pd.DataFrame(rows)
```

## 3. Unsupervised Topic Modeling Comparison (LDA vs. NMF vs. BERTopic c-TF-IDF)

```python
topic_rows = []
for alg_name, info in engine.topic_comparison_data.items():
    topic_rows.append({
        'Topic Model': alg_name,
        'Paradigm': info['algorithm_type'],
        'Coherence Score': info['coherence_score'],
        'Topic Diversity (%)': info['topic_diversity'],
        'Silhouette Score': info['silhouette_score']
    })
pd.DataFrame(topic_rows)
```
